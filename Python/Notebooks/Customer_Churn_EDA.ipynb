```python
def segment_customer(tenure):
    if tenure <= 12:
        return 'New Customer'
    elif tenure <= 48:
        return 'Growing Customer'
    else:
        return 'Loyal Customer'

df['CustomerSegment'] = df['tenure'].apply(segment_customer)
```

```python
df['CustomerSegment'].value_counts()
```

```python
segment_analysis = df.groupby('CustomerSegment').agg(
    Total_Customers=('customerID', 'count'),
    Churned_Customers=('Churn', lambda x: (x == 'Yes').sum()),
    Retained_Customers=('Churn', lambda x: (x == 'No').sum())
)

segment_analysis['Churn_Rate'] = (
    segment_analysis['Churned_Customers'] /
    segment_analysis['Total_Customers'] * 100
)

segment_analysis.round(2)
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='CustomerSegment',
    hue='Churn',
    order=[
        'New Customer',
        'Growing Customer',
        'Loyal Customer'
    ]
)

plt.title('Customer Segmentation by Churn')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=segment_analysis.index,
    y=segment_analysis['Churn_Rate']
)

plt.title('Churn Rate by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Churn Rate (%)')

plt.show()
```

```python
plt.figure(figsize=(15, 5))

for i, column in enumerate(
    ['tenure', 'MonthlyCharges', 'TotalCharges'], 1
):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[column])
    plt.title(f'Outlier Detection - {column}')

plt.tight_layout()
plt.show()
```

```python
numerical_columns = [
    'tenure',
    'MonthlyCharges',
    'TotalCharges'
]

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\nColumn: {column}")
    print("Lower Bound:", round(lower_bound, 2))
    print("Upper Bound:", round(upper_bound, 2))
    print("Number of Outliers:", len(outliers))
```

```python
advanced_analysis = df.groupby(
    ['Contract', 'InternetService']
).agg(
    Total_Customers=('customerID', 'count'),
    Churned_Customers=(
        'Churn',
        lambda x: (x == 'Yes').sum()
    )
)

advanced_analysis['Churn_Rate'] = (
    advanced_analysis['Churned_Customers'] /
    advanced_analysis['Total_Customers'] * 100
)

advanced_analysis.sort_values(
    'Churn_Rate',
    ascending=False
).round(2)
```

# Customer Churn Analysis – Python EDA Conclusions

## 1. Overall Customer Churn

The dataset contains **7,043 customers**. A total of **1,869 customers** have churned, representing an overall churn rate of **26.54%**.

## 2. Contract Analysis

Month-to-month customers have the highest churn rate at **42.71%**, compared with **11.27%** for one-year contracts and **2.83%** for two-year contracts.

## 3. Customer Tenure

New customers (0–12 months) have a churn rate of **47.44%**. Long-term customers (49+ months) have a much lower churn rate of **9.51%**.

## 4. Payment Method

Electronic check customers have the highest observed churn rate at **45.29%**.

## 5. Internet Services

Fiber optic customers have a churn rate of **41.89%**, higher than DSL customers at **18.96%**.

## 6. Customer Demographics

Senior citizens have a churn rate of **41.68%**, compared with **23.61%** for non-senior citizens.

## 7. Additional Services

Customers without Online Security and Tech Support have churn rates of **41.77%** and **41.64%**, respectively.

## 8. Business Recommendations

- Improve onboarding and engagement for new customers.
- Investigate why month-to-month customers leave more frequently.
- Review the payment experience for electronic check customers.
- Evaluate customer satisfaction with fiber optic services.
- Improve awareness and accessibility of support and security services.
- Develop targeted customer retention strategies.

## 9. Conclusion

The analysis identifies customer tenure, contract type, payment method, internet service and additional support services as important factors associated with customer churn.

These findings can help the business develop more informed customer retention strategies.
